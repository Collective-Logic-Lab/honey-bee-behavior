The days with festoons in the 2019 data are roughly 
1) between day 4 and 20
2) between day 46 and 65

To begin, we will focus on day 4, frames 216000 to 217800.  We focus on camera 0 to compare with video data.

The raw video data for day 4 camera 0 (and the top camera focusing on the festooning area) are here: https://edmond.mpg.de/file.xhtml?fileId=237622&version=1.0

```python
import pandas as pd
```

```python
daynum = 4 #52

if daynum > 50:
    datadir = '../Data/220422/zenodo/beetrajectories_days_050_to_085/'
else:
    datadir = '../Data/220422/zenodo/beetrajectories_days_000_to_049/'

df = pd.read_hdf('{}beetrajectories_{:0>3}.hdf'.format(datadir,daynum))
```

```python
# get data restricted to day 4, frames 216000 to 217800, camera 0
```

```python
restricted_data = df[(df['framenum']>=216000) & (df['framenum']<=217800) & (df['camera']==0)]
restricted_data
```

```python
# write to csv file
restricted_data.to_csv('260508_2019_trajectory_data_day4_camera0_frames_216000_to_217800.csv')
```
