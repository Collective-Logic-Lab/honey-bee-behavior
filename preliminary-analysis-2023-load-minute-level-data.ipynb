# load 5-minute data from 2019 dataset

(1-minute data are also available on zenodo)

```python
data_directory = "../Data/240329/df_day5min/"
df = pd.read_hdf(data_directory+'df_day5min_004.hdf')
```

```python
df.columns
```

```python
df[df['Bee unique ID']==1216][['Age','timedivision','Frame 0', 'Frame 1', 'Frame 2', 'Frame 3',
       'Frame 4', 'Frame 5']].dropna()
```

```python
# 2026/5/15 checking bee ages with specific IDs for Shatadru
ages = []
for beeid in range(1216,2047):
    bee_ages = df[df['Bee unique ID']==beeid]['Age']
    if len(bee_ages) > 0:
        ages.append([beeid,bee_ages.iloc[0]])
ages_df = pd.DataFrame(ages,columns=['Bee unique ID','Age'])
```

```python
ages_df.to_csv('260515_2019_day_04_bee_ages_from_5min_data.csv')
```
